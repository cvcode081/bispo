```python
import streamlit as st
import pandas as pd
import numpy as np
import plotly.express as px

# 1. CONFIGURAÇÃO DA PÁGINA (Sempre a primeira chamada st)
st.set_page_config(page_title="Análise de Livros Pro", layout="wide")

# 2. FUNÇÃO DE DADOS COM CACHE (Ciclo de Análise)
@st.cache_data
def processar_dados(file):
    # Entrada: Leitura
    df = pd.read_csv(file, sep=";")

    # Limpeza: Tratar nulos com mediana e remover zeros
    mediana_ano = df["ano"].median()
    df["ano"] = df["ano"].fillna(mediana_ano).astype(int)
    df = df[df["paginas"] > 0].copy()

    # Transformação: Categorização e Década
    df["faixa_paginas"] = df["paginas"].apply(
        lambda x: "Curto" if x < 150 else "Médio" if x <= 350 else "Longo"
    )
    df["decada"] = (df["ano"] // 10) * 10

    return df

# 3. INTERFACE - SIDEBAR
st.sidebar.header("📂 Gestão de Dados")
uploaded_file = st.sidebar.file_uploader("Suba o arquivo livros.csv", type=["csv"])

if uploaded_file:
    df_base = processar_dados(uploaded_file)

    # Filtros Reativos
    st.sidebar.subheader("🔍 Filtros")
    autores = st.sidebar.multiselect("Filtrar Autores", options=df_base["autor"].unique())
    decadas = st.sidebar.multiselect("Filtrar Décadas", options=sorted(df_base["decada"].unique()))

    # Aplicação da lógica de filtro
    df_filtrado = df_base.copy()
    if autores:
        df_filtrado = df_filtrado[df_filtrado["autor"].isin(autores)]
    if decadas:
        df_filtrado = df_filtrado[df_filtrado["decada"].isin(decadas)]

    # 4. DASHBOARD PRINCIPAL
    st.title("📚 Dashboard de Acervo Literário")

    # KPIs com st.metric
    c1, c2, c3 = st.columns(3)
    c1.metric("Total de Livros", len(df_filtrado))
    c2.metric("Média de Páginas", f"{df_filtrado['paginas'].mean():.1f}")
    c3.metric("Autores Únicos", df_filtrado["autor"].nunique())

    # Gráfico Plotly: Páginas por Década
    st.subheader("📊 Evolução do Tamanho dos Livros por Década")
    fig = px.box(df_filtrado, x="decada", y="paginas", color="faixa_paginas",
                 title="Distribuição de Páginas por Década")
    st.plotly_chart(fig, use_container_width=True)

    # Tabela Dinâmica e Dados
    st.subheader("📋 Dados Detalhados")
    st.dataframe(df_filtrado, use_container_width=True)

    # Botão de Exportação
    csv = df_filtrado.to_csv(index=False).encode('utf-8')
    st.download_button("📥 Baixar Dados Filtrados (CSV)", data=csv, file_name="analise_livros.csv")

else:
    st.info("Aguardando upload do arquivo para iniciar a análise.")
```
