```python
import streamlit as st
import pandas as pd
import numpy as np
import plotly.express as px

# 1. CONFIGURAÇÃO DA PÁGINA
st.set_page_config(page_title="Dashboard Pro", layout="wide")

# 2. FUNÇÃO DE DADOS COM CACHE
@st.cache_data
def carregar_dados_exemplo():
    data = {
        "nome": ["Ana", "Bruno", "Carlos", "Daniela", "Eduardo"],
        "idade": [23, 35, 29, 31, 40],
        "cidade": ["SP", "RJ", "SP", "MG", "RJ"],
        "salario": [3000, 5000, 4000, 3500, 4800],
        "categoria_salario": ["Baixo", "Alto", "Médio", "Médio", "Alto"]
    }
    return pd.DataFrame(data)

# 3. LÓGICA DE UPLOAD (Desafio Difícil)
st.sidebar.header("📂 Configurações")
uploaded_file = st.sidebar.file_uploader("Suba seu CSV", type=["csv"])

if uploaded_file:
    df_base = pd.read_csv(uploaded_file)
else:
    df_base = carregar_dados_exemplo()

# 4. FILTROS NA SIDEBAR
st.sidebar.subheader("🔎 Filtros")
cidades = st.sidebar.multiselect("Cidades", options=df_base["cidade"].unique(), default=df_base["cidade"].unique())

# Filtro de Categoria (Desafio Fácil)
categorias = st.sidebar.multiselect("Categorias", options=df_base["categoria_salario"].unique(), default=df_base["categoria_salario"].unique())

# Aplicando Filtros
df_filtrado = df_base[
    (df_base["cidade"].isin(cidades)) &
    (df_base["categoria_salario"].isin(categorias))
]

# 5. DASHBOARD (KPIs e Gráficos)
st.title("📊 Dashboard de Funcionários")

col1, col2 = st.columns(2)
with col1:
    st.metric("Total Funcionários", len(df_filtrado))
with col2:
    st.metric("Média Salarial", f"R$ {df_filtrado['salario'].mean():.2f}")

# Gráfico Plotly (Desafio Médio)
fig = px.bar(df_filtrado, x="nome", y="salario", color="cidade", title="Salários por Funcionário")
st.plotly_chart(fig, use_container_width=True)

st.dataframe(df_filtrado, use_container_width=True)
```
